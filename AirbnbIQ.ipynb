## 4. Data Cleaning & Feature Engineering

Steps:

1. Drop columns that are pure identifiers or free text and carry no generalizable signal for a tabular model (`id`, `name`, `host_id`, `host_name`, `last_review`).
2. Fill missing `reviews_per_month` with `0` (no reviews yet).
3. Cap extreme outliers in `price` and `minimum_nights` using percentile clipping, so a handful of data-entry errors (e.g. $10,000/night, 1,250 minimum nights) don't distort the model.
4. Separate features (`X`) from the target (`y`).

```python
df
```

```python
#kabhi bhi orihinal dataframe me change nhi karte abhi hm ek clone banyenge usme changes kare and jo unwanted dataset hai use hata denge
```

```python
df_clean=df.drop(columns=['id','host_id','name','host_name','last_review'])
df_clean
```

```python
#NO reviews yet-> 0 reviews per month,not missing
df_clean['reviews_per_month']=df_clean['reviews_per_month'].fillna(0)
```

```python
#3 Cap extreme outlier instead of deleting rows.
price_cap=df_clean['price'].quantile(0.99)
nights_cap=df_clean['minimum_nights'].quantile(0.99)

df_clean['price']=df_clean['price'].clip(upper=price_cap)
df_clean['minimum_nights']=df_clean['minimum_nights'].clip(upper=nights_cap)
df_clean

#df_clean=df.clean[(df.clean['price']<price_cap) & (df.clean['minimum_nights']<nights_cap)] ye bhi wahi same clip hi hai
```

```python
X=df_clean.drop(columns='room_type')
y=df_clean['room_type']
```

## 6. Preprocessing: ColumnTransformer + Pipeline

### ColumnTransformer

`ColumnTransformer` is a Scikit-learn tool that allows us to apply different preprocessing techniques to different groups of columns within a single transformer.

For example, numerical and categorical features require different preprocessing, so `ColumnTransformer` helps us handle them separately in one place.

**Why use ColumnTransformer?**
- Applies different transformations to different types of columns.
- Keeps all preprocessing steps organized in one place.
- Ensures the same preprocessing is applied to both training and test data.
- Helps prevent data leakage when used inside a Pipeline.
- Makes preprocessing reusable and easier to maintain.

### Preprocessing Strategy

Production ML code should never manually transform train/test data with separate lines — it's error-prone and can leak information. Instead, we build a single, reusable `ColumnTransformer`:

- **Numeric features** — median imputation + standard scaling
- **Categorical features** — most-frequent imputation + one-hot encoding

This transformer will be the first step of every model pipeline below, so preprocessing is learned only from the training data and applied consistently everywhere (no data leakage).

## 6. Preprocessing: ColumnTransformer + Pipeline

Production ML code should never manually transform train/test data with separate lines — it's error-prone and leaks information. Instead we build a single, reusable `ColumnTransformer`:

- **Numeric features** — median imputation + standard scaling
- **Categorical features** — most-frequent imputation + one-hot encoding

This transformer will be the first step of every model pipeline below, so preprocessing is learned only on training data and applied consistently everywhere (no data leakage).

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
#simpleImputer hmare liye medain and mode aur jo bhi hai wo khud se  fill kar dega
from sklearn.preprocessing import StandardScaler, OneHotEncoder,PowerTransformer
#powerTransformer se hm skewness ko handle kar skate hai
```

```python
numerical_cols = [
    "price",
    "minimum_nights",
    "number_of_reviews",
    "reviews_per_month",
    "calculated_host_listings_count",
    "availability_365",
    "latitude",
    "longitude"
]

```

```python
categorical_cols = [
    "neighbourhood_group",
    "neighbourhood"
]
```

```python
numeric_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
   # ('skewer', PowerTransformer(method='yeo-johnson')),
    ('scaler', StandardScaler())
])
```

```python
#pipeline for categorial
categorical_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    #jo sabse jayda baar aayega usse fill ho jayega
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
    #ignore se agar koi miss hua to use hata denge taki error na aye
])
```

```python
preprocessor=ColumnTransformer(transformers=[
    ('num',numeric_pipeline,numerical_cols),
    ('cat',categorical_pipeline,categorical_cols)
])
#ye bata rha hai ki numeric aayega tab numerical me jayega and catgroy hai to categorical me jayega.
preprocessor
```

## 8. Trying Multiple Algorithms

We now compare several models, each wrapped in its own pipeline with the same preprocessor. We evaluate each with **3-fold stratified cross-validation** on the training set (never touching the test set yet), so the comparison is fair and robust.

### Models tried:

1. **Logistic Regression** — simple, interpretable linear baseline
2. **Decision Tree** — a single non-linear model, prone to overfitting
3. **Random Forest** — an ensemble of trees, usually a strong general-purpose model
4. **Gradient Boosting** — sequential ensemble, often the most accurate of the classics

```python
#we use SMOTE(synthehtic minority oversampling) for class imbalance
#1 it has disadavtage
#High chances of overfitting
#Data leakage

#so hm yaha use karne wale hai class weight , class weight kya karta hai  jo value kam aati hai uspe focus jayda karta hai joshared value aati hai
# usper jayda focus karta hai
#Gradient boosting doesn't support class weight
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import cross_validate
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import warnings
warnings.filterwarnings('ignore')
```

```python
class_weight="Balanced"
#tells the model to pay more attention to the minority class i.e shared rooma but problem hai ki gradient boosting support nhi karta so hm ise rhne dete hai

models={
    'logistic':LogisticRegression(max_iter=1000),
    'decision_tree':DecisionTreeClassifier(),
    'random_forest':RandomForestClassifier(),
    'gradient_boosting':GradientBoostingClassifier()
}

for name,model in models.items():
   pipe=Pipeline(steps=[
       ('preprocessor',preprocessor),
       ('classifier',model)
   ])
 # Accuracy alone can be misleading on imbalanced classes, so we use macro F1
# Both: plain accuracy and macro F1 (which treats every class equally)

accuracy = cross_val_score(pipe, X_train, y_train, cv=3, scoring="accuracy")
macrof1 = cross_val_score(pipe, X_train, y_train, cv=3, scoring="f1_macro")

print(f"{name} -> Accuracy: {accuracy.mean():.3f}, F1: {macrof1.mean():.3f}")
```

```python
models = {
    'logistic': LogisticRegression(max_iter=1000),
    'decision_tree': DecisionTreeClassifier(),
    'random_forest': RandomForestClassifier(),
    'gradient_boosting': GradientBoostingClassifier()
}

for name, model in models.items():

    pipe = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])

    # Accuracy and Macro F1 using 3-fold cross-validation
    accuracy = cross_val_score(
        pipe, X_train, y_train, cv=3, scoring="accuracy"
    )

    macrof1 = cross_val_score(
        pipe, X_train, y_train, cv=3, scoring="f1_macro"
    )

    print(f"{name} -> Accuracy: {accuracy.mean():.3f}, F1: {macrof1.mean():.3f}")
```

### Model Selection

- **Selected Model:** Random Forest Classifier
- Random Forest achieved the **highest Accuracy (85.3%)** among all tested models.
- It also achieved the **highest Macro F1-score (71.6%)**.
- Macro F1 is important because our dataset contains **class imbalance**.
- Random Forest supports `class_weight="balanced"`, which gives more importance to minority classes and helps **reduce bias toward the majority class**.
- Therefore, Random Forest is selected for **hyperparameter tuning**.

## 9. Hyperparameter Tuning

We pick the best-performing model from the comparison above (whichever one that turns out to be) and search over its key hyperparameters using `RandomizedSearchCV` (cheaper than an exhaustive grid search, while still exploring the space well). We keep optimizing for **macro-F1** since the classes are imbalanced.

```python
from sklearn.model_selection import RandomizedSearchCV

best_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(class_weight="balanced", random_state=42))
])

param_distributions = {
    "classifier__n_estimators": [100, 150,200, 250,300],
    "classifier__max_depth": [None, 8,12,15,20],
    "classifier__min_samples_split": [2, 5, 10],
}
search=RandomizedSearchCV(
    best_pipeline,
    param_distributions=param_distributions,
    n_iter=10,
    cv=3,
    scoring="f1_macro",
    random_state=42,
    verbose=1
)
search.fit(X_train, y_train)
```

```python
print("Best Parameters: ",search.best_params_)
print("Best Score: ",search.best_score_)
```

## 10. Final Evaluation on the Test Set

This is the *only time* we touch the held-out test set. It gives us an honest estimate of how the final tuned model would perform on completely unseen listings.

```python
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score

best_pipeline = search.best_estimator_

y_pred = best_pipeline.predict(X_test)

print(f"Accuracy Score: {accuracy_score(y_test, y_pred)}")
print(f"F1 Score: {f1_score(y_test, y_pred, average='macro')}")
```

```python
sns.heatmap(
    confusion_matrix(y_test, y_pred),
    annot=True,
    fmt=".2f",
    xticklabels=best_pipeline.classes_,
    yticklabels=best_pipeline.classes_
)
```

## 11. Model Explainability - Feature Importance

Random Forest provides feature importance scores that help us understand which features contribute the most to the model's predictions.

Higher feature importance indicates that the model relies more heavily on that feature while making predictions.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Get the trained Random Forest model from the pipeline
rf_model = best_pipeline.named_steps["classifier"]

# Get feature names after preprocessing
feature_names = best_pipeline.named_steps[
    "preprocessor"
].get_feature_names_out()

# Create a DataFrame containing feature names and their importance
feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": rf_model.feature_importances_
})

# Sort features from most important to least important
feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

# Display top 15 features
feature_importance.head(15)
```

### Key Observations

- `price` is the most important feature, contributing the most to the model's predictions.
- Location-based features such as `latitude`, `longitude`, and neighbourhood information also influence room-type classification.
- Other listing characteristics such as minimum nights, host listing count, availability, and reviews also contribute to the prediction.
- Overall, the model relies on a combination of pricing, location, and listing characteristics to classify the Airbnb room type.

```python
# Plot top 15 most important features
top_features = feature_importance.head(15)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_features,
    x="Importance",
    y="Feature"
)

plt.title("Top 15 Most Important Features")
plt.xlabel("Feature Importance")
plt.ylabel("Feature")

plt.show()
```
